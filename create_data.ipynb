```python

import pandas as pd
import numpy as np

np.random.seed(42)

n = 200

df = pd.DataFrame({
    'Customer_ID': range(1001, 1001 + n),

    'Age': np.random.randint(18, 70, n),

    'Gender': np.random.choice(
        ['Male', 'Female', 'Other'], n
    ),

    'City': np.random.choice(
        ['Bangalore', 'Mumbai', 'Delhi', 'Chennai',
         'Hyderabad', 'Pune'], n
    ),

    'Signup_Date': pd.date_range(
        start='2023-01-01', periods=n, freq='5D'
    ).strftime('%d/%m/%Y'),

    'Plan': np.random.choice(
        ['Basic', 'Standard', 'Premium'], n
    ),

    'Monthly_Charges': np.random.randint(299, 3000, n),

    'Tenure_Months': np.random.randint(1, 60, n),

    'Support_Tickets': np.random.randint(0, 12, n),

    'Last_Login_Days': np.random.randint(0, 100, n),

    'Payment_Method': np.random.choice(
        ['Credit Card', 'Debit Card', 'UPI', 'Net Banking'], n
    ),

    'Contract_Type': np.random.choice(
        ['Monthly', 'Quarterly', 'Yearly'], n
    ),

    'Monthly_Usage_Hours': np.random.randint(1, 200, n),

    'Satisfaction_Score': np.random.randint(1, 11, n)
})

# Create a target variable with some relationship to customer behaviour
risk = (
    (df['Tenure_Months'] < 12).astype(int) * 2 +
    (df['Support_Tickets'] > 6).astype(int) * 2 +
    (df['Last_Login_Days'] > 45).astype(int) * 2 +
    (df['Satisfaction_Score'] < 5).astype(int) * 2
)

probability = 1 / (1 + np.exp(-(risk - 3)))

df['Churn'] = np.random.binomial(1, probability)

# Introduce missing values
for col in ['Age', 'City', 'Monthly_Charges',
            'Payment_Method', 'Satisfaction_Score']:
    idx = np.random.choice(df.index, 5, replace=False)
    df.loc[idx, col] = np.nan

# Introduce empty strings
df.loc[[10, 25, 40], 'City'] = ''
df.loc[[15, 35, 55], 'Payment_Method'] = ''

# Introduce inconsistent text formatting
df.loc[5, 'Gender'] = ' male '
df.loc[12, 'Plan'] = 'basic'
df.loc[20, 'City'] = 'bangalore'
df.loc[30, 'Payment_Method'] = 'upi'

# Introduce outliers
df.loc[50, 'Age'] = 150
df.loc[60, 'Monthly_Charges'] = 50000
df.loc[70, 'Monthly_Usage_Hours'] = 1000

# Introduce duplicate records
df.loc[198] = df.loc[100]
df.loc[199] = df.loc[120]

# Save dataset
df.to_csv('Customer_Churn_200.csv', index=False)

print("Dataset created successfully!")
print("Shape:", df.shape)
print(df.head())
```
