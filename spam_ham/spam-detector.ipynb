```python
data["v1"] = data["v1"].replace({
    "ham": 0,
    "spam": 1
})
```

```python
data.head()
```

```python
import nltk

nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
from nltk.tokenize import word_tokenize
data["tokens"] = data["v2"].fillna("").astype(str).apply(word_tokenize)
```

```python
data['tokens'].head()
```

```python
data["tokens"] = data["tokens"].apply(
    lambda tokens: [token.lower() for token in tokens]
)
```

```python
print("Missing token rows:", data["tokens"].isna().sum())
print("Token column dtype:", data["tokens"].dtype)
data["tokens"].head()
```
