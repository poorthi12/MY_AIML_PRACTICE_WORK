```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_excel('flight_price.xlsx')
```

```python
df.head(5)
```

```python
df['Journey_day']=pd.to_datetime(df['Date_of_Journey']).dt.day
```

```python
df['Journey_month'] = pd.to_datetime(df['Date_of_Journey']).dt.month
```

```python
df['Journey_year'] = pd.to_datetime(df['Date_of_Journey']).dt.year
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df['Dep_Hour']=df['Dep_Time'].str.split(':').str[0]
```

```python
df['Dep_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df['Arrival_Hour']=df['Arrival_Time'].str.split(':').str[0]
```

```python
df['Arrival_Minutes']=df['Arrival_Time'].str.split(':').str[1].str.split(' ').str[0]
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df['Duration_Hour']=df['Duration'].str.split('h').str[0]
```

```python
df['Duration_Min']=df['Duration'].str.split('h').str[-1].str.replace('m'," ").str.strip()
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
df.Total_Stops.unique()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0, '2 stops':2, '1 stop':1, '3 stops':3, 'nan':np.nan, '4 stops':4})
```

```python
df.drop('Route',axis=1,inplace=True)
```
