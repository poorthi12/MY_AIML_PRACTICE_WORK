```python
columns = ['sex','class','who','embark_town',]
```

```python
from sklearn.preprocessing import OneHotEncoder
one = OneHotEncoder(sparse_output=False)
encoded = one.fit_transform(df[columns])
```

```python
encoded
```

```python
encoded_data=pd.DataFrame(encoded,columns=one.get_feature_names_out(columns),index=df.index)
```

```python
data = pd.concat([df,encoded_data],axis=1)
```

```python
data
```

```python
data.columns
```

```python
data.drop(data[columns],axis=1,inplace=True)
```

```python
data
```

```python
data.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data['alive'] = le.fit_transform(data['alive'])
```

```python
data
```

```python
data.info()
```

```python
corelation = data.corr(numeric_only=True)
```

```python
data.drop(columns=['sex_male','class_First','class_Second','class_Third','who_man','who_woman'],inplace=True)
```

```python
corelation['alive'].sort_values(ascending=False)
```

```python
data.drop_duplicates(inplace=True)
```

```python
corelation['alive'].sort_values(ascending=False)
```

```python
x = data.drop(['parch','sibsp','age','alive'],axis=1)
```

```python
x
```

```python
y = data['alive']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=200,max_depth=5,random_state=42)
model.fit(x_train_scaled,y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
